```python
import os
import requests

# -------------------------------------------------------------------------
# GitHub Details
# -------------------------------------------------------------------------
github_user = "sachin0249"
repo_name = "UIDAI---FABRIC-RAW-DATA"
folder_path = "raw"  # GitHub source folder path
branch = "main"

# GitHub REST API with recursive=1 to fetch full directory tree including subfolders
api_url = f"https://api.github.com/repos/{github_user}/{repo_name}/git/trees/{branch}?recursive=1"

# Base destination in Lakehouse Files section
base_destination = "/lakehouse/default/Files/raw"

# -------------------------------------------------------------------------
# Download all nested CSV files
# -------------------------------------------------------------------------
response = requests.get(api_url)

if response.status_code == 200:
    tree = response.json().get("tree", [])
    downloaded_count = 0
    
    for item in tree:
        path = item.get("path", "")
        
        # Check if item is inside the 'raw' folder and is a CSV file
        if path.startswith(f"{folder_path}/") and item.get("type") == "blob" and path.lower().endswith(".csv"):
            # Construct raw file URL on GitHub
            download_url = f"https://raw.githubusercontent.com/{github_user}/{repo_name}/{branch}/{path}"
            
            # Map GitHub 'raw/...' path to target 'raw_ingestion/...' folder inside Lakehouse
            relative_subpath = os.path.relpath(path, folder_path)
            local_file_path = os.path.join(base_destination, relative_subpath)
            
            # Create subfolders dynamically if they don't exist
            os.makedirs(os.path.dirname(local_file_path), exist_ok=True)
            
            print(f"Downloading: {path} -> {local_file_path}")
            file_data = requests.get(download_url).content
            
            with open(local_file_path, 'wb') as f:
                f.write(file_data)
                
            downloaded_count += 1
            
    print(f"\nSuccessfully downloaded {downloaded_count} CSV file(s) maintaining subfolder paths under: {base_destination}")
else:
    print(f"Failed to access GitHub API. HTTP Status: {response.status_code}")
```
